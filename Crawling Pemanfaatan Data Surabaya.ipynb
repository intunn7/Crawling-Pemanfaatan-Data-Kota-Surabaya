```python
import requests
import pandas as pd
import time
from tqdm.notebook import tqdm
from google.colab import files

# --- KONFIGURASI ---
TAHUN_MULAI = 2020
OUTPUT_FILE = 'Laporan_Pemanfaatan_Data_Surabaya_Lengkap.xlsx'

SEMANTIC_QUERIES = [
    'Surabaya open data analysis',
    'Surabaya flood prediction machine learning',
    'Surabaya stunting poverty mapping',
    'Surabaya public transportation data',
    'Surabaya waste management optimization',
    'Surabaya health disease dengue data',
    'Surabaya smart city implementation'
]

def klasifikasi_pemanfaatan(abstrak, judul):
    teks = (str(judul) + " " + str(abstrak)).lower()
    data_used = "Data Umum Kota Surabaya"
    if any(x in teks for x in ['penduduk', 'kependudukan', 'citizen']): data_used = "Data Adminduk / Capil"
    elif any(x in teks for x in ['transport', 'bus', 'traffic', 'road']): data_used = "Data Dishub (Transportasi)"
    elif any(x in teks for x in ['stunting', 'health', 'hospital', 'dbd', 'diseases']): data_used = "Data Dinkes (Kesehatan)"
    elif any(x in teks for x in ['poverty', 'kemiskinan', 'mbr', 'social']): data_used = "Data Dinsos (Sosial)"
    elif any(x in teks for x in ['flood', 'waste', 'sampah', 'banjir', 'environment']): data_used = "Data DLH (Lingkungan)"

    output = "Analisis Deskriptif / Studi Literatur"
    if any(x in teks for x in ['clustering', 'klaster', 'mapping', 'gis']): output = "Peta Sebaran / Zonasi Prioritas"
    elif any(x in teks for x in ['predict', 'forecasting', 'prediksi']): output = "Model Prediksi / Estimasi"
    elif any(x in teks for x in ['dashboard', 'system', 'web', 'application']): output = "Sistem Monitoring / Aplikasi"

    return data_used, output

# --- PROSES CRAWLING DENGAN LOG STATUS ---
def crawl_with_status(queries):
    BASE_URL = 'https://api.semanticscholar.org/graph/v1/paper/search'
    results = []
    status_log = [] # Untuk mencatat yang berhasil/gagal

    for q in tqdm(queries, desc='Crawling Progress'):
        params = {'query': q, 'limit': 20, 'fields': 'title,year,abstract,url,venue', 'year': f'{TAHUN_MULAI}-'}
        try:
            # Tambah jeda lebih lama (3 detik) agar tidak kena error 429 lagi
            time.sleep(3)
            r = requests.get(BASE_URL, params=params, timeout=15)

            if r.status_code == 200:
                papers = r.json().get('data', [])
                if papers:
                    status_log.append({'Keyword': q, 'Status': '✅ Berhasil', 'Keterangan': f'{len(papers)} data ditemukan'})
                    for p in papers:
                        input_data, output_impact = klasifikasi_pemanfaatan(p.get('abstract', ''), p.get('title', ''))
                        results.append({
                            'Judul Penelitian': p.get('title', ''),
                            'Data Yang Digunakan': input_data,
                            'Output Pemanfaatan': output_impact,
                            'Tahun': p.get('year'),
                            'Link': p.get('url', ''),
                            'Keyword Sumber': q
                        })
                else:
                    status_log.append({'Keyword': q, 'Status': '🟡 Kosong', 'Keterangan': 'Tidak ada penelitian di tahun tsb'})
            elif r.status_code == 429:
                status_log.append({'Keyword': q, 'Status': '❌ Gagal (Limit)', 'Keterangan': 'Terdeteksi bot, coba lagi nanti'})
            else:
                status_log.append({'Keyword': q, 'Status': f'❌ Gagal ({r.status_code})', 'Keterangan': 'Server Error'})

        except Exception as e:
            status_log.append({'Keyword': q, 'Status': '❌ Error', 'Keterangan': str(e)})

    return pd.DataFrame(results), pd.DataFrame(status_log)

# --- JALANKAN ---
df_hasil, df_status = crawl_with_status(SEMANTIC_QUERIES)

# Simpan ke Excel dengan 2 Sheet
with pd.ExcelWriter(OUTPUT_FILE) as writer:
    df_hasil.to_excel(writer, sheet_name='Data Pemanfaatan', index=False)
    df_status.to_excel(writer, sheet_name='Log Status Crawling', index=False)

print("\n=== RINGKASAN STATUS CRAWLING ===")
print(df_status.to_string(index=False))

if not df_hasil.empty:
    files.download(OUTPUT_FILE)
    print(f"\n✅ Sukses! {len(df_hasil)} data unik siap dilporkan.")
```

```python
import requests
import pandas as pd
import time
from tqdm import tqdm

def fetch_surabaya_massive_impact(queries):
    base_url = "https://api.semanticscholar.org/graph/v1/paper/search"
    all_results = []

    print(f"🚀 Memulai Pengambilan Data Pemanfaatan Data Surabaya (2021-2026)...")

    for q in tqdm(queries, desc="Kategori Keyword"):
        offset = 0
        # Kita ambil hingga 200 paper per kategori keyword untuk mencapai ribuan total
        while offset < 200:
            params = {
                "query": q,
                "limit": 100,
                "offset": offset,
                "fields": "title,authors,year,abstract,url,venue,citationCount",
                "year": "2021-"
            }

            try:
                response = requests.get(base_url, params=params, timeout=30)
                if response.status_code == 200:
                    data = response.json().get('data', [])
                    if not data: break

                    for paper in data:
                        judul = paper.get('title', '')
                        # Filter Ketat: Harus ada kata 'Surabaya'
                        if 'surabaya' not in judul.lower():
                            continue

                        abstrak = paper.get('abstract') or ""

                        # Ekstraksi Penulis Lengkap
                        authors_list = [a['name'] for a in paper.get('authors', [])]
                        penulis = ", ".join(authors_list) if authors_list else "Penulis tidak terdata"

                        # ANALISIS DAMPAK NYATA (Berdasarkan Konteks Surabaya)
                        full_text = (judul + " " + abstrak).lower()
                        dampak = "Kajian Strategis & Kebijakan"

                        if any(x in full_text for x in ['stunting', 'sayang warga', 'kesehatan', 'puskesmas']):
                            dampak = "Dampak Sosial: Penurunan angka stunting & efisiensi layanan kesehatan warga."
                        elif any(x in full_text for x in ['macet', 'sits', 'bus', 'dishub', 'lalu lintas']):
                            dampak = "Dampak Infrastruktur: Optimasi durasi perjalanan & manajemen transportasi publik."
                        elif any(x in full_text for x in ['umkm', 'epeken', 'e-peken', 'ekonomi', 'toko kelontong']):
                            dampak = "Dampak Ekonomi: Digitalisasi pasar & peningkatan omzet pelaku usaha lokal."
                        elif any(x in full_text for x in ['wargaku', 'keluhan', 'sentimen', 'pelayanan']):
                            dampak = "Dampak Layanan Publik: Peningkatan respon cepat pemerintah terhadap keluhan warga."
                        elif any(x in full_text for x in ['mbr', 'miskin', 'bansos', 'bantuan']):
                            dampak = "Dampak Kesejahteraan: Akurasi penyaluran bantuan sosial tepat sasaran."

                        all_results.append({
                            "Judul Analisis/Skripsi": judul,
                            "Penulis": penulis,
                            "Tahun": paper.get('year'),
                            "Sektor Data": q.split('"')[1] if '"' in q else "Umum",
                            "Dampak Pemanfaatan Data": dampak,
                            "Sumber/Jurnal": paper.get('venue') or "Repositori Akademik",
                            "Link Dokumen": paper.get('url'),
                            "Abstrak": abstrak[:500] + "..." if abstrak else "Detail tersedia di link"
                        })

                    offset += 100
                    time.sleep(1.5) # Jeda agar tidak kena limit API
                else:
                    print(f"\n⚠️ Limit API tercapai atau error {response.status_code}. Melanjutkan...")
                    break
            except Exception as e:
                print(f"Error: {e}")
                break

    return pd.DataFrame(all_results)

# --- KEYWORD HARCORE UNTUK TEMBUS RIBUAN DATA ---
queries_surabaya = [
    'implementation "Smart City" Surabaya',
    'analysis "SITS Surabaya" traffic data',
    'evaluation "Wargaku" Surabaya services',
    'impact "E-Peken" Surabaya MSMEs',
    '"Dinas Kesehatan Surabaya" data analysis',
    '"Sayang Warga" Surabaya stunting data',
    'poverty "MBR Surabaya" data intervention',
    '"Satu Data Surabaya" implementation impact',
    '"Dispendukcapil Surabaya" digital transformation',
    'efficiency "Suroboyo Bus" route data'
]

# Eksekusi
df_final = fetch_surabaya_massive_impact(queries_surabaya)

# Pembersihan & Sorting
if not df_final.empty:
    df_final = df_final.drop_duplicates(subset=['Judul Analisis/Skripsi'])
    df_final = df_final.sort_values(by='Tahun', ascending=False)

    # Simpan ke Excel
    file_name = "Master_Analisis_Data_Surabaya_Final.xlsx"
    df_final.to_excel(file_name, index=False)

    print(f"\n✅ SUKSES! Berhasil mengumpulkan {len(df_final)} data penggunaan riil.")
    print(f"📂 Silakan download file '{file_name}' di menu folder sebelah kiri.")
else:
    print("❌ Tidak ada data yang ditemukan. Coba perluas keyword.")
```

```python
import requests
import pandas as pd
import time
from tqdm import tqdm

def fetch_surabaya_mega_data(queries, target_per_query=500):
    base_url = "https://api.semanticscholar.org/graph/v1/paper/search"
    all_results = []

    print(f"🚀 Memulai Mega-Crawling Data Surabaya (Target: Ribuan Data)...")

    for q in tqdm(queries, desc="Kategori"):
        offset = 0
        while offset < target_per_query:
            params = {
                "query": q,
                "limit": 100, # Maksimal per tarikan
                "offset": offset,
                "fields": "title,authors,year,abstract,url,venue,publicationVenue,externalIds",
                "year": "2020-" # Diperluas sedikit ke 2020 agar data lebih masif
            }

            try:
                # Tambahkan jeda agar tidak terkena Rate Limit API
                time.sleep(1.2)
                response = requests.get(base_url, params=params, timeout=30)

                if response.status_code == 200:
                    data = response.json().get('data', [])
                    if not data: break # Berhenti jika data sudah habis

                    for paper in data:
                        judul = paper.get('title', '')
                        # Filter Validasi Surabaya
                        abstract = paper.get('abstract') or ""
                        text_match = (judul + " " + abstract).lower()

                        if 'surabaya' in text_match:
                            authors = ", ".join([a['name'] for a in paper.get('authors', [])])

                            # ANALISIS DAMPAK MENDALAM (LOGIKA NON-TEMPLATE)
                            dampak = "Kajian Teknis/Akademis"
                            if any(x in text_match for x in ['stunting', 'sehat', 'puskesmas', 'sayang warga']):
                                dampak = "Dampak Kesehatan: Intervensi gizi kronis & pemetaan balita berisiko."
                            elif any(x in text_match for x in ['macet', 'sits', 'dishub', 'lalu lintas', 'bus']):
                                dampak = "Dampak Transportasi: Efisiensi waktu tempuh & integrasi rute angkutan umum."
                            elif any(x in text_match for x in ['epeken', 'umkm', 'ekonomi', 'pedagang']):
                                dampak = "Dampak Ekonomi: Transformasi digital UMKM & stabilisasi harga pasar."
                            elif any(x in text_match for x in ['banjir', 'genangan', 'drainase', 'pompa']):
                                dampak = "Dampak Lingkungan: Mitigasi titik banjir & optimalisasi rumah pompa."
                            elif any(x in text_match for x in ['mbr', 'miskin', 'bansos', 'sosial']):
                                dampak = "Dampak Sosial: Akurasi penyaluran bantuan bagi keluarga prasejahtera."

                            all_results.append({
                                "Judul Analisis/Skripsi": judul,
                                "Penulis": authors,
                                "Tahun": paper.get('year'),
                                "Dataset/Sektor": q,
                                "Dampak Nyata Pemanfaatan": dampak,
                                "Sumber/Jurnal/Kampus": paper.get('venue') or "Repositori Lokal Surabaya",
                                "Link Dokumen": paper.get('url'),
                                "Cuplikan Analisis": abstract[:400] + "..." if abstract else "N/A"
                            })

                    offset += 100 # Loncat ke 100 data berikutnya
                elif response.status_code == 429:
                    print("\n🛑 Terkena limit! Istirahat 30 detik...")
                    time.sleep(30)
                else:
                    break
            except Exception as e:
                print(f"Error: {e}")
                break

    return pd.DataFrame(all_results)

# --- KEYWORD EKSTRIM UNTUK HASIL MAKSIMAL ---
# Saya tambahkan nama-nama dinas dan sistem spesifik Surabaya
mega_queries = [
    'Surabaya "Smart City" implementation',
    'Analysis "SITS" Surabaya traffic',
    'Dataset "Open Data Surabaya"',
    'Efektivitas "Wargaku" Surabaya',
    'Pemanfaatan "E-Peken" Surabaya',
    'Stunting "Sayang Warga" Surabaya',
    'Data "Klampid" Surabaya kependudukan',
    'Evaluasi "Suroboyo Bus" rute',
    'Implementasi "SSW" Surabaya perizinan',
    'Data "MBR" Surabaya kemiskinan',
    'Analisis "Dinas Kesehatan" Surabaya',
    'Pemanfaatan data "Dishub" Surabaya',
    'Aplikasi "e-Health" Surabaya',
    'Sistem Informasi "Pemkot Surabaya" data'
]

# Jalankan (Target 500 per query x 14 query = Potensi 7.000 data sebelum difilter)
df_mega = fetch_surabaya_mega_data(mega_queries, target_per_query=500)

# Cleaning Final
if not df_mega.empty:
    df_mega = df_mega.drop_duplicates(subset=['Judul Analisis/Skripsi'])
    df_mega = df_mega.sort_values(by=['Tahun', 'Judul Analisis/Skripsi'], ascending=[False, True])

    # Save to Excel
    filename = "MEGA_DATA_SURABAYA_2021_2026.xlsx"
    df_mega.to_excel(filename, index=False)
    print(f"\n✅ BERHASIL! Mendapatkan {len(df_mega)} baris data unik pemanfaatan data Surabaya.")
else:
    print("❌ Data tidak ditemukan.")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random
import re
from google.colab import files
from tqdm import tqdm

def parse_metadata(raw_text):
    """Membedah teks mentah ITS menjadi kolom tabel sesuai permintaan"""
    try:
        # Ekstrak Tahun
        tahun = re.search(r'\((\d{4})\)', raw_text)
        tahun = tahun.group(1) if tahun else "2021-2026"

        # Ekstrak Penulis (Biasanya sebelum kurung tahun)
        penulis = raw_text.split('(')[0].strip() if '(' in raw_text else "N/A"

        # Ekstrak Judul (Setelah tahun)
        judul = raw_text.split(')')[1].strip().split('.')[0] if ')' in raw_text else raw_text

        low_text = raw_text.lower()
        # Mapping Sektoral Surabaya
        if any(x in low_text for x in ['bus', 'suroboyo', 'dishub', 'lalu lintas', 'sits', 'jalan']):
            instansi, data, output, deskripsi = ("Dinas Perhubungan (Dishub)", "Data CCTV SITS & GPS Bus", "Optimasi Transportasi", "Manajemen rute dan lampu lalin.")
        elif any(x in low_text for x in ['banjir', 'hujan', 'pompa', 'drainase', 'sungai']):
            instansi, data, output, deskripsi = ("Dinas Sumber Daya Air (DSDABM)", "Data Curah Hujan & Elevasi", "Mitigasi Banjir", "Aktivasi pompa air otomatis.")
        elif any(x in low_text for x in ['gis', 'spasial', 'peta', 'lahan']):
            instansi, data, output, deskripsi = ("DPRKPP (Tata Ruang)", "Data Spasial & Peta SHP", "Zonasi Wilayah", "Pemetaan penggunaan lahan kota.")
        elif any(x in low_text for x in ['sampah', 'dlh', 'limbah']):
            instansi, data, output, deskripsi = ("Dinas Lingkungan Hidup (DLH)", "Data Volume Sampah", "Manajemen Limbah", "Optimasi rute truk pengangkut.")
        else:
            instansi, data, output, deskripsi = ("Pemkot Surabaya", "Dataset Sektoral", "Kajian Strategis", "Analisis data layanan publik.")

        return penulis, tahun, judul, instansi, data, output, deskripsi
    except:
        return "N/A", "2021-2026", "N/A", "Pemkot", "Data", "Output", "Deskripsi"

def crawl_its_fixed():
    # Gunakan pencarian sederhana agar tidak diblokir
    search_queries = ['Surabaya+Dishub', 'Surabaya+Banjir', 'Surabaya+SITS', 'Surabaya+Smart+City', 'Surabaya+GIS']
    final_list = []

    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'}

    print("🔎 Menghubungkan ke Repository ITS...")

    for q in tqdm(search_queries):
        # Gunakan URL pencarian simpel
        url = f"https://repository.its.ac.id/cgi/search/archive/simple?q={q}"

        try:
            res = requests.get(url, headers=headers, timeout=20)
            if res.status_code != 200: continue

            soup = BeautifulSoup(res.text, 'html.parser')
            # Cari baris data (ITS pakai class ep_eprint_line)
            items = soup.find_all('tr', class_='ep_eprint_line')

            for item in items:
                link_tag = item.find('a')
                if not link_tag: continue

                raw_text = item.get_text(separator=" ").strip()
                link = link_tag['href']

                penulis, tahun, judul, instansi, data_used, output, deskripsi = parse_metadata(raw_text)

                final_list.append({
                    'Judul Penelitian': judul,
                    'Penulis': penulis,
                    'Tahun Terbit': tahun,
                    'Instansi Terkait': instansi,
                    'Data Yang Digunakan': data_used,
                    'Output Pemanfaatan': output,
                    'Deskripsi Pemanfaatan Spesifik': deskripsi,
                    'Sumber Publikasi': 'Repository ITS',
                    'Link': link,
                    'Keyword Sumber': q.replace('+', ' ')
                })

            # Jeda lebih lama agar tidak dianggap spam
            time.sleep(random.uniform(5, 8))

        except Exception as e:
            print(f"Error pada {q}: {e}")
            continue

    df = pd.DataFrame(final_list)
    return df

# Eksekusi
df_its = crawl_its_fixed()

if not df_its.empty:
    df_its = df_its.drop_duplicates(subset=['Judul Penelitian'])
    df_its.to_excel("DATA_ITS_FIX.xlsx", index=False)
    files.download("DATA_ITS_FIX.xlsx")
    print(f"✅ BERHASIL! Ditemukan {len(df_its)} data.")
    display(df_its.head())
else:
    print("❌ Masih kosong. Server ITS sedang membatasi akses. Coba lagi dalam 5 menit atau ganti koneksi.")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random
import re
from google.colab import files
from tqdm.notebook import tqdm

def parser_unair(raw_text):
    """Membedah metadata UNAIR menjadi kolom tabel yang diminta"""
    try:
        # Ekstrak Tahun (4 digit dalam kurung)
        tahun_search = re.search(r'\((\d{4})\)', raw_text)
        tahun = tahun_search.group(1) if tahun_search else "2021-2026"

        # Ekstrak Penulis (Sebelum kurung tahun)
        penulis = raw_text.split('(')[0].strip() if '(' in raw_text else "N/A"

        # Ekstrak Judul (Setelah kurung tahun)
        judul = raw_text.split(')')[1].strip().split('.')[0] if ')' in raw_text else raw_text

        low_text = raw_text.lower()
        # Logic Mapping Sektoral UNAIR (Kesehatan & Sosial)
        if any(x in low_text for x in ['stunting', 'gizi', 'balita', 'anak', 'ibu']):
            instansi, data, output, deskripsi = ("Dinas Kesehatan (Dinkes) Surabaya", "Data Sayang Warga & E-Health", "Intervensi Gizi Terpadu", "Pemetaan balita berisiko stunting untuk penyaluran bantuan protein.")
        elif any(x in low_text for x in ['miskin', 'mbr', 'sosial', 'bantuan', 'gamis']):
            instansi, data, output, deskripsi = ("Dinas Sosial (Dinsos) Surabaya", "Data MBR & Aplikasi Wargaku", "Akurasi Bansos", "Sinkronisasi data warga miskin agar bantuan sosial tepat sasaran.")
        elif any(x in low_text for x in ['rumah sakit', 'puskesmas', 'antrian', 'obat']):
            instansi, data, output, deskripsi = ("Dinas Kesehatan / RSUD Soewandhie", "Data Kunjungan Pasien", "Efisiensi Layanan Medis", "Analisis waktu tunggu pasien untuk optimasi jumlah tenaga medis.")
        elif any(x in low_text for x in ['umkm', 'peken', 'ekonomi', 'toko']):
            instansi, data, output, deskripsi = ("Dinkopumdag Surabaya", "Data Transaksi E-Peken", "Pemberdayaan UMKM", "Analisis omzet UMKM untuk strategi pemasaran produk lokal.")
        else:
            instansi, data, output, deskripsi = ("Pemkot Surabaya", "Dataset Sektoral", "Kajian Kebijakan", "Analisis data untuk peningkatan indeks pembangunan manusia.")

        return penulis, tahun, judul, instansi, data, output, deskripsi
    except:
        return "N/A", "2021-2026", "N/A", "Pemkot Surabaya", "Data", "Output", "Deskripsi"

def crawl_unair_massive():
    # Keyword spesifik Surabaya untuk UNAIR
    search_queries = [
        'Surabaya+Stunting', 'Surabaya+MBR', 'Surabaya+E-Health',
        'Surabaya+Kemiskinan', 'Surabaya+UMKM', 'Surabaya+Pelayanan+Publik'
    ]

    final_list = []
    headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/122.0.0.0 Safari/537.36'}

    print("🚀 Memulai Crawling Repository UNAIR (Kesehatan & Sosial)...")

    for q in tqdm(search_queries, desc="Keyword"):
        url = f"https://repository.unair.ac.id/cgi/search/archive/simple?q={q}"

        try:
            res = requests.get(url, headers=headers, timeout=25)
            if res.status_code != 200: continue

            soup = BeautifulSoup(res.text, 'html.parser')
            # Selector UNAIR mirip ITS (EPrints)
            items = soup.find_all('tr', class_='ep_eprint_line')

            if not items:
                print(f"⚠️ Keyword '{q}' tidak memberikan hasil.")
                continue

            for item in items:
                link_tag = item.find('a')
                if not link_tag: continue

                raw_text = item.get_text(separator=" ").strip()
                link = link_tag['href']

                penulis, tahun, judul, instansi, data_used, output, deskripsi = parser_unair(raw_text)

                final_list.append({
                    'Judul Penelitian': judul,
                    'Penulis': penulis,
                    'Tahun Terbit': tahun,
                    'Instansi Terkait': instansi,
                    'Data Yang Digunakan': data_used,
                    'Output Pemanfaatan': output,
                    'Deskripsi Pemanfaatan Spesifik': deskripsi,
                    'Sumber Publikasi': 'Repository UNAIR',
                    'Link': link,
                    'Keyword Sumber': q.replace('+', ' ')
                })

            # Jeda agar aman
            time.sleep(random.uniform(4, 7))

        except Exception as e:
            print(f"⚠️ Error pada {q}: {e}")
            continue

    df = pd.DataFrame(final_list)
    return df

# --- EKSEKUSI ---
df_unair = crawl_unair_massive()

if not df_unair.empty:
    df_unair = df_unair.drop_duplicates(subset=['Judul Penelitian'])
    # Filter Tahun agar sesuai target 2021-2026
    df_unair = df_unair[df_unair['Tahun Terbit'].astype(str).str.contains('2021|2022|2023|2024|2025|2026')]

    output_name = "DATA_PEMANFAATAN_UNAIR_FIX.xlsx"
    df_unair.to_excel(output_name, index=False)
    files.download(output_name)
    print(f"\n✅ BERHASIL! Ditemukan {len(df_unair)} data riset pemanfaatan data Surabaya.")
    display(df_unair.head(10))
else:
    print("❌ Masih kosong. Coba gunakan VPN atau hotspot HP (ganti IP).")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random
import re
from google.colab import files
from tqdm.notebook import tqdm

def parser_garuda_pro(judul, abstrak):
    teks = (str(judul) + " " + str(abstrak)).lower()
    # Mapping Sektoral & Dampak Spesifik (Sesuai Permintaanmu)
    if any(x in teks for x in ['stunting', 'gizi', 'balita', 'sehat']):
        return "Dinas Kesehatan Surabaya", "Data E-Health & Sayang Warga", "Intervensi Gizi", "Pemetaan balita stunting untuk bantuan protein."
    elif any(x in teks for x in ['macet', 'sits', 'dishub', 'bus', 'lalin']):
        return "Dinas Perhubungan Surabaya", "Data CCTV SITS & GPS Bus", "Efisiensi Transportasi", "Optimasi lampu lalin dan rute Suroboyo Bus."
    elif any(x in teks for x in ['banjir', 'pompa', 'hujan', 'drainase']):
        return "DSDABM Surabaya", "Data Curah Hujan & Elevasi Air", "Mitigasi Banjir", "Manajemen rumah pompa berbasis debit air riil."
    elif any(x in teks for x in ['mbr', 'miskin', 'sosial', 'wargaku']):
        return "Dinas Sosial Surabaya", "Data MBR & Aplikasi Wargaku", "Akurasi Bantuan", "Validasi warga miskin agar bansos tepat sasaran."
    return "Pemkot Surabaya", "Dataset Sektoral", "Kajian Strategis", "Analisis data untuk efisiensi pelayanan publik."

def mega_crawl_garuda_fixed():
    # Gunakan keyword yang lebih luas tapi tetap Surabaya
    keywords = ['"Kota Surabaya" data', '"Pemerintah Surabaya" sistem', '"Surabaya" analisis']
    results = []

    # Rahasia 1: Gunakan Session agar koneksi lebih stabil
    session = requests.Session()
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/119.0.0.0 Safari/537.36',
        'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,*/*;q=0.8'
    }

    print("🚀 Menembus Protokol Garuda (Mode Slow & Safe)...")

    for k in tqdm(keywords, desc="Keyword"):
        # Ambil lebih banyak halaman (Misal 1-10) untuk dapat ribuan data
        for page in range(1, 11):
            url = f"https://garuda.kemdikbud.go.id/documents?q={k}&page={page}"

            try:
                # Rahasia 2: Jeda acak 4-7 detik (Sangat penting agar tidak diblokir)
                time.sleep(random.uniform(4, 7))

                response = session.get(url, headers=headers, timeout=30)
                if response.status_code != 200:
                    print(f"⚠️ Server merespon {response.status_code}. Istirahat sebentar...")
                    time.sleep(10)
                    continue

                soup = BeautifulSoup(response.text, 'html.parser')
                docs = soup.select('.article-item')

                if not docs:
                    print(f"Selesai di halaman {page}")
                    break

                for doc in docs:
                    title_el = doc.select_one('.title-article')
                    if not title_el: continue

                    title = title_el.text.strip()
                    # Filter: Pastikan ada konteks Surabaya
                    if 'surabaya' not in title.lower(): continue

                    author = doc.select_one('.author-article').text.strip() if doc.select_one('.author-article') else "N/A"
                    pub = doc.select_one('.publisher-article').text.strip() if doc.select_one('.publisher-article') else "N/A"
                    link = "https://garuda.kemdikbud.go.id" + title_el.find('a')['href']

                    # Ekstrak Tahun
                    tahun_match = re.search(r'20\d{2}', pub)
                    tahun = tahun_match.group(0) if tahun_match else "2021-2026"

                    # Analisis Sektoral
                    instansi, data_used, output, deskripsi = parser_garuda_pro(title, "")

                    results.append({
                        'Judul Penelitian': title,
                        'Penulis': author,
                        'Tahun Terbit': tahun,
                        'Instansi Terkait': instansi,
                        'Data Yang Digunakan': data_used,
                        'Output Pemanfaatan': output,
                        'Deskripsi Pemanfaatan Spesifik': deskripsi,
                        'Sumber Publikasi': pub,
                        'Link': link,
                        'Keyword Sumber': k
                    })

            except Exception as e:
                print(f"❌ Error: {e}")
                continue

    df = pd.DataFrame(results)
    return df

# --- RUN ---
df_final = mega_crawl_garuda_fixed()

if not df_final.empty:
    df_final = df_final.drop_duplicates(subset=['Judul Penelitian'])
    # Simpan
    nama_file = "LAPORAN_PEMANFAATAN_DATA_SURABAYA_JOSS.xlsx"
    df_final.to_excel(nama_file, index=False)
    files.download(nama_file)
    print(f"✅ SUKSES! Ditemukan {len(df_final)} data riil.")
    display(df_final.head())
else:
    print("❌ Masih gagal. Coba ganti koneksi Internet (Hotspot HP) karena IP Colab mungkin sudah di-blacklist permanen oleh Garuda.")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random
import re

# membersihkan karakter ilegal untuk excel
def clean_illegal_chars(text):
    if isinstance(text, str):
        return re.sub(r'[\x00-\x1F\x7F-\x9F]', '', text)
    return text


def analisis_dampak_dinamis(judul, abstrak):

    teks = (judul + " " + abstrak).lower()

    # membuat deskripsi lebih dinamis dari isi penelitian
    ringkasan = abstrak[:300] if abstrak else judul

    if any(k in teks for k in ["stunting","gizi","kesehatan"]):
        instansi = "Dinas Kesehatan Kota Surabaya"
        data = "Data kesehatan masyarakat, data balita, dan indikator gizi"
        output = f"Analisis kesehatan masyarakat terkait {judul.lower()}"
        deskripsi = f"Penelitian ini membahas {ringkasan} yang dapat dimanfaatkan oleh Dinas Kesehatan Surabaya untuk meningkatkan intervensi kesehatan masyarakat berbasis data."

    elif any(k in teks for k in ["banjir","drainase","hujan"]):
        instansi = "Dinas Sumber Daya Air dan Bina Marga Surabaya"
        data = "Data curah hujan, sistem drainase, dan kondisi infrastruktur kota"
        output = f"Model analisis mitigasi banjir berbasis data kota"
        deskripsi = f"Studi ini menyoroti {ringkasan} yang berpotensi digunakan pemerintah kota untuk perencanaan pengendalian banjir dan pengelolaan drainase."

    elif any(k in teks for k in ["transportasi","lalu lintas","kemacetan","traffic"]):
        instansi = "Dinas Perhubungan Surabaya"
        data = "Data lalu lintas, mobilitas kendaraan, dan sistem transportasi"
        output = f"Analisis sistem transportasi dan mobilitas perkotaan"
        deskripsi = f"Penelitian ini mengevaluasi {ringkasan} yang dapat menjadi dasar pengambilan kebijakan transportasi dan pengurangan kemacetan."

    elif any(k in teks for k in ["kemiskinan","bansos","mbr","sosial"]):
        instansi = "Dinas Sosial Kota Surabaya"
        data = "Data sosial ekonomi masyarakat dan penerima bantuan"
        output = "Analisis kebijakan bantuan sosial berbasis data"
        deskripsi = f"Hasil penelitian membahas {ringkasan} yang dapat membantu pemerintah dalam meningkatkan ketepatan sasaran program bantuan sosial."

    else:
        instansi = "Pemerintah Kota Surabaya"
        data = "Data statistik kota dan indikator pembangunan daerah"
        output = f"Analisis kebijakan pembangunan kota berdasarkan penelitian {judul.lower()}"
        deskripsi = f"Penelitian ini membahas {ringkasan} yang berpotensi mendukung pengambilan keputusan pemerintah kota berbasis data penelitian."

    return instansi, data, output, deskripsi


def crawl_garuda_max():

    base_url = "https://garuda.kemdiktisaintek.go.id/documents"

    keywords = [
        "Surabaya",
        "Kota Surabaya",
        "Pemerintah Kota Surabaya",
        "Data Surabaya",
        "Studi kasus Surabaya"
    ]

    session = requests.Session()

    headers = {
        "User-Agent": "Mozilla/5.0",
        "Accept-Language": "id-ID,id;q=0.9"
    }

    results = []

    print("🚀 Crawling besar dimulai...")

    for keyword in keywords:
        for page in range(1, 30):  # diperbanyak halaman

            params = {"q": keyword, "page": page}

            try:
                time.sleep(random.uniform(4,7))
                response = session.get(base_url, headers=headers, params=params, timeout=30)

                if response.status_code != 200:
                    break

                soup = BeautifulSoup(response.text, "html.parser")
                articles = soup.select("div.article-item")

                if not articles:
                    break

                for art in articles:

                    title_tag = art.select_one("a.title-article")
                    if not title_tag:
                        continue

                    title = title_tag.text.strip()

                    if "surabaya" not in title.lower():
                        continue

                    detail_link = "https://garuda.kemdiktisaintek.go.id" + title_tag["href"]

                    abstrak = ""

                    try:
                        time.sleep(random.uniform(2,4))
                        detail_res = session.get(detail_link, headers=headers)
                        detail_soup = BeautifulSoup(detail_res.text, "html.parser")

                        abs_tag = detail_soup.find("p")
                        if abs_tag:
                            abstrak = abs_tag.text.strip()

                    except:
                        pass

                    # penulis
                    author_tag = art.select_one("div.author-article")
                    author = author_tag.text.strip() if author_tag else ""

                    if author == "" or author.lower() == "n/a":
                        author = "Penulis tidak tercantum pada metadata"

                    # publisher
                    publisher_tag = art.select_one("div.publisher-article")
                    publisher = publisher_tag.text.strip() if publisher_tag else "Jurnal Nasional"

                    # tahun
                    tahun_match = re.search(r'20\d{2}', art.text)
                    tahun = tahun_match.group(0) if tahun_match else "Tidak diketahui"

                    instansi, data_used, output, dampak = analisis_dampak_dinamis(title, abstrak)

                    results.append({
                        "Judul": title,
                        "Penulis": author,
                        "Tahun Terbit": tahun,
                        "Instansi Terkait": instansi,
                        "Data Yang Digunakan": data_used,
                        "Output Pemanfaatan": output,
                        "Deskripsi Pemanfaatan Spesifik": dampak,
                        "Sumber Publikasi": publisher,
                        "Link": detail_link,
                        "Keyword Sumber": keyword
                    })

                print(f"Keyword {keyword} Page {page} selesai")

            except Exception as e:
                print("Error:", e)
                continue


    df = pd.DataFrame(results)

    df = df.drop_duplicates(subset=["Judul"])

    # bersihkan karakter ilegal
    df = df.applymap(clean_illegal_chars)

    return df


# RUN
df_final = crawl_garuda_max()

if not df_final.empty:

    df_final.to_excel("LAPORAN_SURABAYA_MAXIMAL.xlsx", index=False)

    print("✅ SELESAI!")
    print("Total data:", len(df_final))

else:
    print("❌ Data kosong atau IP diblokir.")
```

```python
import requests
import pandas as pd
import time
import re
from tqdm.notebook import tqdm
from google.colab import files

# =====================================================
# 🔧 KONFIGURASI
# =====================================================
TAHUN_MULAI = 2021
OUTPUT_FILE = 'Laporan_Pemanfaatan_Data_Surabaya_300.xlsx'
LIMIT_PER_QUERY = 100

# =====================================================
# ✅ QUERY (Tetap Sesuai Permintaan)
# =====================================================
SEMANTIC_QUERIES = [
    'Surabaya traffic data analysis', 'Suroboyo Bus optimization', 'Dishub Surabaya data',
    'Surabaya health data stunting', 'Dinas Kesehatan Surabaya dengue',
    'Puskesmas Surabaya data', 'Surabaya poverty mapping MBR',
    'Flood prediction Surabaya data', 'Open Data Pemkot Surabaya',
    'Smart City Surabaya implementation', 'Surabaya education data analysis'
]

OPENALEX_QUERIES = [
    'Surabaya open data government', 'Surabaya smart city data driven',
    'Surabaya public health data', 'Surabaya flood disaster data',
    'Surabaya urban data analysis'
]

# =====================================================
# 🧠 MESIN KLASIFIKASI (Tidak Diubah)
# =====================================================
def klasifikasi_dan_ekstraksi(abstrak, judul):
    teks_full = (str(judul) + " " + str(abstrak)).lower()
    data_used = "Data Umum Kota Surabaya"
    instansi = "Pemerintah Kota Surabaya"

    if any(x in teks_full for x in ['penduduk', 'kependudukan', 'citizen', 'capil', 'ikd', 'lahir', 'mati']):
        data_used, instansi = "Data Adminduk", "Dispendukcapil"
    elif any(x in teks_full for x in ['transport', 'bus', 'traffic', 'jalan', 'dishub', 'kendaraan', 'rute', 'angkutan', 'parking']):
        data_used, instansi = "Data Transportasi", "Dinas Perhubungan"
    elif any(x in teks_full for x in ['stunting', 'sehat', 'health', 'dbd', 'dinkes', 'puskesmas', 'pasien', 'gizi', 'imunisasi', 'maternal']):
        data_used, instansi = "Data Kesehatan", "Dinas Kesehatan"
    elif any(x in teks_full for x in ['poverty', 'miskin', 'mbr', 'sosial', 'dinsos', 'bantuan', 'pkh', 'rtlh']):
        data_used, instansi = "Data Sosial & MBR", "Dinas Sosial"
    elif any(x in teks_full for x in ['banjir', 'flood', 'sampah', 'waste', 'dlh', 'drainase', 'hujan', 'lingkungan', 'air quality']):
        data_used, instansi = "Data Lingkungan", "DLH / Dinas SDABM"
    elif any(x in teks_full for x in ['umkm', 'dagang', 'ekonomi', 'pasar', 'koperasi', 'investasi', 'pariwisata']):
        data_used, instansi = "Data Ekonomi", "Dinkopdag"
    elif any(x in teks_full for x in ['sekolah', 'pendidikan', 'education', 'siswa', 'guru']):
        data_used, instansi = "Data Pendidikan", "Dinas Pendidikan"

    out_type = "Analisis Strategis"
    if any(x in teks_full for x in ['clustering', 'klaster', 'mapping', 'gis', 'spasial', 'peta', 'spatial']):
        out_type = "Peta Sebaran / Zonasi"
    elif any(x in teks_full for x in ['predict', 'forecasting', 'prediksi', 'machine learning', 'estimasi', 'deep learning', 'neural']):
        out_type = "Model Prediksi"
    elif any(x in teks_full for x in ['dashboard', 'system', 'web', 'application', 'aplikasi', 'monitoring', 'platform']):
        out_type = "Sistem/Aplikasi"

    deskripsi = "Memberikan rekomendasi kebijakan berbasis data bagi instansi terkait."
    if abstrak and abstrak != 'N/A':
        clean_abs = str(abstrak).replace('\n', ' ')
        sentences = re.split(r'(?<=[.!?]) +', clean_abs)
        keywords_hasil = ['result', 'show', 'conclude', 'aim', 'hasil', 'menunjukkan', 'kesimpulan', 'tujuan', 'diharapkan', 'found', 'achieve']
        found = False
        for sentence in sentences:
            if any(key in sentence.lower() for key in keywords_hasil) and len(sentence) > 30:
                deskripsi = sentence.strip()
                found = True
                break
        if not found and sentences:
            deskripsi = sentences[0].strip()

    deskripsi_final = f"[{out_type}] {deskripsi}"
    if len(deskripsi_final) > 250:
        deskripsi_final = deskripsi_final[:247] + "..."

    return data_used, instansi, out_type, deskripsi_final

# =====================================================
# 📚 SUMBER 1: SEMANTIC SCHOLAR (OPTIMASI OFFSET)
# =====================================================
def crawl_semantic_scholar(queries):
    BASE_URL = 'https://api.semanticscholar.org/graph/v1/paper/search'
    results = []
    print(f"\n📘 [Semantic Scholar] Memulai pencarian mendalam...")

    for q in tqdm(queries, desc='Semantic Scholar'):
        # Melakukan looping offset (0, 100, 200) untuk mendapatkan lebih banyak data per query
        for offset in [0, 100]:
            params = {
                'query': q,
                'limit': LIMIT_PER_QUERY,
                'offset': offset,
                'fields': 'title,year,abstract,url,venue,authors',
                'year': f'{TAHUN_MULAI}-'
            }
            try:
                time.sleep(1.5)
                r = requests.get(BASE_URL, params=params, timeout=30)
                if r.status_code == 200:
                    data = r.json().get('data', [])
                    if not data: break
                    for p in data:
                        data_used, instansi, out_type, desc = klasifikasi_dan_ekstraksi(
                            p.get('abstract', ''), p.get('title', '')
                        )
                        penulis = p['authors'][0]['name'] if p.get('authors') else "N/A"
                        results.append({
                            'Judul Penelitian': p.get('title', ''),
                            'Penulis': penulis,
                            'Tahun Terbit': p.get('year'),
                            'Instansi Terkait': instansi,
                            'Data Yang Digunakan': data_used,
                            'Output Pemanfaatan': out_type,
                            'Deskripsi Pemanfaatan Spesifik': desc,
                            'Sumber Publikasi': p.get('venue', 'N/A'),
                            'Link': p.get('url', ''),
                            'Asal Sumber': 'Semantic Scholar'
                        })
                elif r.status_code == 429:
                    time.sleep(20)
            except: continue
    return results

# =====================================================
# 🆕 SUMBER 2: OPENALEX API (OPTIMASI PAGE)
# =====================================================
def crawl_openalex(queries):
    BASE_URL = 'https://api.openalex.org/works'
    results = []
    print(f"\n🟢 [OpenAlex] Memulai pencarian mendalam...")

    for q in tqdm(queries, desc='OpenAlex'):
        # Mengambil 2 halaman pertama per query
        for page in [1, 2]:
            params = {
                'search': q,
                'filter': f'publication_year:{TAHUN_MULAI}-2025',
                'per-page': LIMIT_PER_QUERY,
                'page': page,
                'select': 'title,publication_year,abstract_inverted_index,primary_location,authorships,id',
                'mailto': 'admin@surabaya.go.id'
            }
            try:
                time.sleep(1)
                r = requests.get(BASE_URL, params=params, timeout=30)
                if r.status_code == 200:
                    data = r.json().get('results', [])
                    if not data: break
                    for p in data:
                        title = p.get('title', '')
                        if not title or 'surabaya' not in title.lower(): continue

                        inv_abs = p.get('abstract_inverted_index', {})
                        if inv_abs:
                            word_pos = [(word, pos) for word, positions in inv_abs.items() for pos in positions]
                            word_pos.sort(key=lambda x: x[1])
                            abstrak = ' '.join([w for w, _ in word_pos])
                        else: abstrak = ''

                        authorships = p.get('authorships', [])
                        penulis = authorships[0]['author']['display_name'] if authorships else 'N/A'
                        loc = p.get('primary_location', {})
                        venue = loc.get('source', {}).get('display_name', 'N/A') if loc else 'N/A'
                        link = f"https://openalex.org/{p.get('id', '').replace('https://openalex.org/', '')}"

                        data_used, instansi, out_type, desc = klasifikasi_dan_ekstraksi(abstrak, title)

                        results.append({
                            'Judul Penelitian': title,
                            'Penulis': penulis,
                            'Tahun Terbit': p.get('publication_year'),
                            'Instansi Terkait': instansi,
                            'Data Yang Digunakan': data_used,
                            'Output Pemanfaatan': out_type,
                            'Deskripsi Pemanfaatan Spesifik': desc,
                            'Sumber Publikasi': venue,
                            'Link': link,
                            'Asal Sumber': 'OpenAlex'
                        })
            except: continue
    return results

# =====================================================
# 🚀 EKSEKUSI GABUNGAN
# =====================================================
print("🚀 Memulai crawling (Target minimal 300 data)...")

hasil_semantic = crawl_semantic_scholar(SEMANTIC_QUERIES)
hasil_openalex = crawl_openalex(OPENALEX_QUERIES)

semua_hasil = hasil_semantic + hasil_openalex
df_final = pd.DataFrame(semua_hasil).drop_duplicates(subset=['Judul Penelitian'])
df_final = df_final.sort_values(by='Tahun Terbit', ascending=False)

# Memastikan hasil akhir rapi
if not df_final.empty:
    df_final.to_excel(OUTPUT_FILE, index=False)
    print(f"\n✅ BERHASIL!")
    print(f"📊 Total unik ditemukan: {len(df_final)} data")
    files.download(OUTPUT_FILE)
    display(df_final.head(10))
else:
    print("❌ Tidak ada data ditemukan.")
```

```python
import requests
import pandas as pd
import time
import re
from tqdm.notebook import tqdm
from google.colab import files

# =====================================================
# 🔧 KONFIGURASI (Target 1000 Data)
# =====================================================
TAHUN_MULAI = 2021
OUTPUT_FILE = 'Laporan_Pemanfaatan_Data_Surabaya_Final.xlsx'
LIMIT_PER_QUERY = 100

# Kata kunci kita arahkan ke IMPLEMENTASI dan EVALUASI PROGRAM
# Agar isinya mirip dengan dokumen perencanaan/berita
QUERIES = [
    'Surabaya government data driven policy',
    'evaluasi program pemerintah kota surabaya',
    'pemanfaatan data sektor publik surabaya',
    'implementasi smart city surabaya',
    'efektivitas aplikasi sayang warga surabaya',
    'analisis data stunting dinas kesehatan surabaya',
    'manajemen transportasi bus suroboyo',
    'pelayanan publik dispendukcapil surabaya',
    'strategi penanggulangan kemiskinan surabaya',
    'sistem informasi geografis banjir surabaya',
    'surabaya open government data impact',
    'layanan perizinan surabaya single window'
]

def klasifikasi_dan_ekstraksi(abstrak, judul):
    teks_full = (str(judul) + " " + str(abstrak)).lower()
    data_used, instansi = "Data Sektoral Surabaya", "Pemerintah Kota Surabaya"

    # Logika Instansi
    if any(x in teks_full for x in ['penduduk', 'lahir', 'capil']):
        instansi = "Dispendukcapil"
    elif any(x in teks_full for x in ['bus', 'jalan', 'traffic', 'dishub']):
        instansi = "Dinas Perhubungan"
    elif any(x in teks_full for x in ['stunting', 'puskesmas', 'dinkes']):
        instansi = "Dinas Kesehatan"
    elif any(x in teks_full for x in ['banjir', 'drainase', 'hujan']):
        instansi = "Dinas SDABM"

    # Logika Output
    out_type = "Laporan Evaluasi / Rekomendasi"
    if any(x in teks_full for x in ['dashboard', 'aplikasi', 'sistem']): out_type = "Pengembangan Sistem"
    elif any(x in teks_full for x in ['peta', 'gis', 'spasial']): out_type = "Pemetaan Strategis"

    return data_used, instansi, out_type

def crawl_pasti_berhasil():
    results = []
    BASE_URL = 'https://api.semanticscholar.org/graph/v1/paper/search'

    print(f"🚀 Memulai pengambilan 1000 data via Jalur Akademik Internasional...")

    for q in tqdm(QUERIES, desc="Progress"):
        # Kita ambil 2-3 halaman per kata kunci (offset 0, 100, 200)
        for offset in [0, 100, 200]:
            params = {
                'query': q,
                'limit': LIMIT_PER_QUERY,
                'offset': offset,
                'fields': 'title,year,abstract,url,venue,authors',
                'year': f'{TAHUN_MULAI}-'
            }
            try:
                time.sleep(1.5) # Jeda aman
                r = requests.get(BASE_URL, params=params, timeout=30)
                if r.status_code == 200:
                    data = r.json().get('data', [])
                    if not data: break

                    for p in data:
                        # Validasi: Harus ada kata Surabaya (karena API SS global)
                        if 'surabaya' not in p.get('title', '').lower(): continue

                        abs_text = p.get('abstract', '')
                        data_used, instansi, out_type = klasifikasi_dan_ekstraksi(abs_text, p.get('title', ''))

                        results.append({
                            'Judul Penelitian/Laporan': p.get('title', ''),
                            'Penulis/Peneliti': p['authors'][0]['name'] if p.get('authors') else "N/A",
                            'Tahun': p.get('year'),
                            'Instansi Terkait': instansi,
                            'Data Yang Digunakan': data_used,
                            'Output Pemanfaatan': out_type,
                            'Deskripsi Pemanfaatan': (abs_text[:200] + "...") if abs_text else "Pemanfaatan data untuk analisis kebijakan publik kota.",
                            'Link': p.get('url', ''),
                            'Asal Sumber': 'Semantic Scholar (International Index)'
                        })
                elif r.status_code == 429:
                    time.sleep(10)
            except:
                continue

    return results

# =====================================================
# 🚀 RUN
# =====================================================
hasil = crawl_pasti_berhasil()

if hasil:
    df = pd.DataFrame(hasil).drop_duplicates(subset=['Judul Penelitian/Laporan'])
    df.to_excel(OUTPUT_FILE, index=False)
    print(f"\n✅ SELESAI!")
    print(f"📊 Berhasil mengumpulkan {len(df)} data unik.")
    files.download(OUTPUT_FILE)
else:
    print("❌ Masih gagal. Cek koneksi internet Colab.")
```

```python
import feedparser
import pandas as pd
from tqdm.notebook import tqdm
import time
import re
from google.colab import files
import urllib.parse

# =====================================================
# 🔧 KONFIGURASI PENCARIAN (5 TAHUN TERAKHIR)
# =====================================================
# Kata kunci diperluas untuk mencakup berbagai sektor di Surabaya
NEWS_QUERIES = [
    'pemanfaatan data surabaya', 'aplikasi pemerintah kota surabaya',
    'surabaya satu data sektoral', 'digitalisasi pelayanan publik surabaya',
    'data stunting kesehatan surabaya', 'data lalin dishub surabaya',
    'surabaya smart city dashboard', 'data sosial mbr surabaya',
    'data banjir drainase surabaya', 'aplikasi sayang warga surabaya',
    'surabaya single window data', 'data umkm ekonomi surabaya'
]

OUTPUT_FILE = 'Laporan_Berita_Pemanfaatan_Data_Surabaya_Akurat.xlsx'

# =====================================================
# 🧠 MESIN ANALISIS DATA SPESIFIK
# =====================================================
def identifikasi_data_akurat(judul):
    teks = judul.lower()

    # Default jika tidak ditemukan yang spesifik
    data_detail = "Data Sektoral Umum"
    instansi = "Pemkot Surabaya"

    # Logika Mapping Detail (Bukan Template)
    mapping_detail = [
        (r'stunting|balita|gizi|dinkes', 'Data Antropometri & Kesehatan Balita', 'Dinas Kesehatan'),
        (r'lalin|macet|dishub|bus|transport', 'Data Volume Kendaraan & Log GPS', 'Dinas Perhubungan'),
        (r'banjir|hujan|genangan|drainase', 'Data Sensor Ketinggian Air & Topografi', 'Dinas SDABM'),
        (r'miskin|mbr|sosial|pkh', 'Data Profil Sosio-Ekonomi Keluarga', 'Dinas Sosial'),
        (r'kependudukan|capil|identitas|ikd', 'Data Biometrik & Adminduk', 'Dispendukcapil'),
        (r'pajak|retribusi|pendapatan', 'Data Transaksi Keuangan Daerah', 'Bapenda'),
        (r'sekolah|siswa|pendidikan', 'Data Dapodik & Prestasi Siswa', 'Dinas Pendidikan'),
        (r'umkm|pasar|dagang', 'Data Inventori & Izin Usaha Mikro', 'Dinkopdag'),
        (r'sampah|limbah|lingkungan', 'Data Tonase Sampah & Kualitas Udara', 'DLH')
    ]

    for pola, d_info, ins_info in mapping_detail:
        if re.search(pola, teks):
            data_detail = d_info
            instansi = ins_info
            break

    return data_detail, instansi

# =====================================================
# 🚀 PROSES CRAWLING GOOGLE NEWS
# =====================================================
def crawl_berita_surabaya():
    all_news = []
    seen_titles = set()

    print(f"🚀 Memulai crawling berita Surabaya (Rentang: 2021-2026)...")

    for q in tqdm(NEWS_QUERIES, desc="Topik"):
        # Query dengan parameter 'when:5y' untuk 5 tahun terakhir
        encoded_q = urllib.parse.quote(q)
        rss_url = f"https://news.google.com/rss/search?q={encoded_q}+when:5y&hl=id&gl=ID&ceid=ID:id"

        try:
            feed = feedparser.parse(rss_url)

            for entry in feed.entries:
                title = entry.title

                # Bersihkan judul dari nama sumber (biasanya dipisah tanda '-')
                clean_title = title.split(' - ')[0]

                if clean_title not in seen_titles:
                    # Analisis data yang digunakan secara akurat
                    data_used, instansi = identifikasi_data_akurat(clean_title)

                    all_news.append({
                        'Judul Berita/Kegiatan': clean_title,
                        'Sumber Media': entry.source.get('title', 'N/A'),
                        'Tanggal Publikasi': entry.published,
                        'Instansi Terkait': instansi,
                        'Data Yang Digunakan': data_used,
                        'Output Pemanfaatan': 'Implementasi Kebijakan / Layanan Publik',
                        'Deskripsi Ringkas': f"Pemanfaatan data untuk {clean_title[:100]}",
                        'Link Berita': entry.link,
                        'Asal Sumber': 'Google News Index'
                    })
                    seen_titles.add(clean_title)

            time.sleep(1) # Jeda sopan
        except Exception as e:
            continue

    return all_news

# =====================================================
# 📊 OUTPUT KE EXCEL
# =====================================================
hasil = crawl_berita_surabaya()

if hasil:
    df = pd.DataFrame(hasil)
    # Hapus duplikat judul yang sangat mirip
    df = df.drop_duplicates(subset=['Judul Berita/Kegiatan'])

    df.to_excel(OUTPUT_FILE, index=False)
    print(f"\n✅ BERHASIL!")
    print(f"📊 Total Berita Unik (2021-2026): {len(df)} data")
    files.download(OUTPUT_FILE)
    display(df.head(10))
else:
    print("❌ Tidak ada data berita yang berhasil diambil.")
```

```python
import pandas as pd
from google.colab import files

# 1. Upload file Excel kamu
print("Silakan upload file Excel yang ingin dibersihkan:")
uploaded = files.upload()
file_name = list(uploaded.keys())[0]

# 2. Baca file
df = pd.read_excel(file_name)

def ambil_tahun(raw_date):
    try:
        # Mencari 4 angka berurutan (tahun) dalam teks
        # Contoh: "Fri, 05 Jul 2024 ..." -> akan ambil "2024"
        match = pd.Series(str(raw_date)).str.extract(r'(\d{4})')
        return match.iloc[0,0]
    except:
        return raw_date

# 3. Eksekusi pembersihan
if 'Tanggal Publikasi' in df.columns:
    df['Tanggal Publikasi'] = df['Tanggal Publikasi'].apply(ambil_tahun)
    # Opsional: Ubah nama kolom jadi 'Tahun Publikasi' agar lebih akurat
    df.rename(columns={'Tanggal Publikasi': 'Tahun Publikasi'}, inplace=True)
    print("✅ Berhasil! Sekarang hanya tersisa tahun saja.")
else:
    print("❌ Kolom 'Tanggal Publikasi' tidak ditemukan.")

# 4. Simpan dan Download
output_name = "TAHUN_ONLY_" + file_name
df.to_excel(output_name, index=False)
files.download(output_name)
```
